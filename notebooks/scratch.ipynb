### Push Dataset To Huggung Face Hub

```python
from dotenv import load_dotenv
from datasets import load_from_disk

load_dotenv()

DATASET_PATH = "/mnt/huggingface/data/distilled/pmc_patients"
DATASET_NAME = "pmc-patients-distilled-medgemma-22B"
PATH = f"{DATASET_PATH}/{DATASET_NAME}"
HF_REPO = f"loknezmonzter/{DATASET_NAME}"

# Load from disk and push to hub
dataset = load_from_disk(PATH)
print(type(dataset))
```

```python
print(dataset)
```

```python
c10 = 0
c09 = 0
c08 = 0
c07 = 0
c06 = 0
c05 = 0
c00 = 0
c0 = 0

for record in dataset['train'].select(range(len(dataset['train']))):
    rc = len(record['data']['relationships'])
    if rc == 10:
        c10 += 1
    if rc == 9:
        c09 += 1
    if rc == 8:
        c08 += 1
    if rc == 7:
        c07 += 1
    if rc == 6:
        c06 += 1
    if rc == 5:
        c05 += 1
    if rc > 0 and rc < 5:
        c00 += 1
    if rc == 0:
        c0 += 1

d = {
    "10_count": c10,
    "9_count": c09,
    "8_count": c08,
    "7_count": c07,
    "6_count": c06,
    "5_count": c05,
    "1to4_count": c00,
    "0_count": c0
}
```
